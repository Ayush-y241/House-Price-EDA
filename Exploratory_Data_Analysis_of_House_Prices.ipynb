```python
columns = ['SalePrice', 'GrLivArea', 'LotArea']

for col in columns:
    print("="*60)
    print(f"Statistics for {col}")
    print("="*60)

    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode()[0])
    print("Standard Deviation:", df[col].std())
    print("Variance:", df[col].var())
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Range:", df[col].max() - df[col].min())

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    print("IQR:", IQR)
    print("Skewness:", df[col].skew())
    print("Kurtosis:", df[col].kurt())
    print("\n")
```

```python
for col in columns:
    mean = df[col].mean()
    median = df[col].median()

    print(f"\n{col}")
    print("Mean :", mean)
    print("Median :", median)

    if mean > median:
        print("→ Right Skewed Distribution")
    elif mean < median:
        print("→ Left Skewed Distribution")
    else:
        print("→ Approximately Symmetric")
```

```python
def iqr_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return series[(series < lower) | (series > upper)]
```

```python
from scipy.stats import zscore

def zscore_outliers(series, threshold=3):
    z = np.abs(zscore(series))
    return series[z > threshold]
```

```python
columns = ["GrLivArea", "SalePrice"]

for col in columns:
    iqr = iqr_outliers(df[col])
    z = zscore_outliers(df[col])

    print("="*50)
    print(col)
    print("IQR Outliers :", len(iqr))
    print("Z-score Outliers :", len(z))
```
